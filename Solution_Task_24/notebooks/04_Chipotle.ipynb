# Ex1 - Filtering and Sorting Data

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv(data_file('chipotle.tsv'), sep='\t')
chipo.head()
chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).str.strip().astype(float)
chipo['unit_price'] = chipo['item_price'] / chipo['quantity']
```

### Step 4. How many products cost more than $10.00?

```python
# Count distinct named products with at least one unit price above $10.
chipo.loc[chipo['unit_price'].gt(10), 'item_name'].nunique()
```

### Step 5. What is the price of each item? 
###### print a data frame with only two columns item_name and item_price

```python
chipo[['item_name', 'unit_price']].drop_duplicates().rename(columns={'unit_price': 'item_price'}).sort_values(['item_name', 'item_price'])
```

### Step 6. Sort by the name of the item

```python
chipo.sort_values('item_name')
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
chipo.loc[chipo['item_price'].eq(chipo['item_price'].max()), ['item_name', 'quantity', 'item_price']]
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo['item_name'].eq('Veggie Salad Bowl').sum()
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
(chipo['item_name'].eq('Canned Soda') & chipo['quantity'].gt(1)).sum()
```
