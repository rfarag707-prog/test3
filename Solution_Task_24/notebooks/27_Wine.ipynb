# Wine

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data). 

### Step 3. Assign it to a variable called wine

```python
wine = pd.read_csv(data_file('wine.data'), header=None)
wine.head()
```

### Step 4. Delete the first, fourth, seventh, nineth, eleventh, thirteenth and fourteenth columns

```python
wine = wine.drop(columns=wine.columns[[0, 3, 6, 8, 10, 12, 13]])
wine
```

### Step 5. Assign the columns as below:

The attributes are (donated by Riccardo Leardi, riclea '@' anchem.unige.it):  
1) alcohol  
2) malic_acid  
3) alcalinity_of_ash  
4) magnesium  
5) flavanoids  
6) proanthocyanins  
7) hue 

```python
wine.columns = ['alcohol', 'malic_acid', 'alcalinity_of_ash', 'magnesium', 'flavanoids', 'proanthocyanins', 'hue']
wine
```

### Step 6. Set the values of the first 3 rows from alcohol as NaN

```python
wine.loc[wine.index[:3], 'alcohol'] = np.nan
```

### Step 7. Now set the value of the rows 3 and 4 of magnesium as NaN

```python
wine.loc[[3, 4], 'magnesium'] = np.nan
```

### Step 8. Fill the value of NaN with the number 10 in alcohol and 100 in magnesium

```python
wine = wine.fillna({'alcohol': 10, 'magnesium': 100})
```

### Step 9. Count the number of missing values

```python
wine.isna().sum()
```

### Step 10.  Create an array of 10 random numbers up until 10

```python
rng = np.random.default_rng(42)
random_rows = rng.integers(0, 10, size=10)
random_rows
```

### Step 11.  Use random numbers you generated as an index and assign NaN value to each of cell.

```python
wine.loc[random_rows, :] = np.nan
```

### Step 12.  How many missing values do we have?

### Step 13. Delete the rows that contain missing values

```python
wine = wine.dropna()
wine
```

```python
wine.loc[wine['alcohol'].notna(), 'alcohol']
```

```python
wine = wine.reset_index(drop=True)
wine
```
