# Occupation

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv(data_file('u.user'), sep='|')
users.head()
```

### Step 4. Discover what is the mean age per occupation

```python
users.groupby('occupation')['age'].mean()
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
users.assign(is_male=users['gender'].eq('M')).groupby('occupation')['is_male'].mean().sort_values(ascending=False)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
users.groupby('occupation')['age'].agg(['min', 'max'])
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
users.groupby(['occupation', 'gender'])['age'].mean()
```

### Step 8.  For each occupation present the percentage of women and men

```python
pd.crosstab(users['occupation'], users['gender'], normalize='index').mul(100)
```
