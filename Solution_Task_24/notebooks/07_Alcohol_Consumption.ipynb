# Ex - GroupBy

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

### Step 3. Assign it to a variable called drinks.

```python
drinks = pd.read_csv(data_file('drinks.csv'), keep_default_na=False)
drinks.head()
```

### Step 4. Which continent drinks more beer on average?

```python
beer = drinks.groupby('continent')['beer_servings'].mean()
beer[beer.eq(beer.max())]
```

### Step 5. For each continent print the statistics for wine consumption.

```python
drinks.groupby('continent')['wine_servings'].describe()
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
drinks.groupby('continent').mean(numeric_only=True)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
drinks.groupby('continent').median(numeric_only=True)
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
drinks.groupby('continent')['spirit_servings'].agg(['mean', 'min', 'max'])
```
