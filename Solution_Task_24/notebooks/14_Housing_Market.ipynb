### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Create 3 differents Series, each of length 100, as follows: 
1. The first a random number from 1 to 4 
2. The second a random number from 1 to 3
3. The third a random number from 10,000 to 30,000

```python
rng = np.random.default_rng(42)
s1 = pd.Series(rng.integers(1, 5, 100))
s2 = pd.Series(rng.integers(1, 4, 100))
s3 = pd.Series(rng.integers(10000, 30001, 100))
```

### Step 3. Let's create a DataFrame by joinning the Series by column

```python
housing = pd.concat([s1, s2, s3], axis=1)
housing
```

### Step 4. Change the name of the columns to bedrs, bathrs, price_sqr_meter

```python
housing.columns = ['bedrs', 'bathrs', 'price_sqr_meter']
housing
```

### Step 5. Create a one column DataFrame with the values of the 3 Series and assign it to 'bigcolumn'

```python
bigcolumn = pd.concat([s1, s2, s3]).to_frame(name='value')
bigcolumn
```
