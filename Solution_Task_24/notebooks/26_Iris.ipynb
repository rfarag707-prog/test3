# Iris

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data). 

### Step 3. Assign it to a variable called iris

```python
iris = pd.read_csv(data_file('iris.data'), header=None)
iris.head()
```

### Step 4. Create columns for the dataset

```python
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
iris
```

### Step 5.  Is there any missing value in the dataframe?

```python
iris.isna().any().any()
```

### Step 6.  Lets set the values of the rows 10 to 29 of the column 'petal_length' to NaN

```python
iris.loc[10:29, 'petal_length'] = np.nan
```

### Step 7. Good, now lets substitute the NaN values to 1.0

```python
iris['petal_length'] = iris['petal_length'].fillna(1.0)
```

### Step 8. Now let's delete the column class

```python
iris = iris.drop(columns='class')
```

### Step 9.  Set the first 3 rows as NaN

```python
iris.iloc[:3, :] = np.nan
```

### Step 10.  Delete the rows that have NaN

```python
iris = iris.dropna()
iris
```

### Step 11. Reset the index so it begins with 0 again

```python
iris = iris.reset_index(drop=True)
iris
```
