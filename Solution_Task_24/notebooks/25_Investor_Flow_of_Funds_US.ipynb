# Investor - Flow of Funds - US

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/datasets/investor-flow-of-funds-us/master/data/weekly.csv). 

### Step 3. Assign it to a variable called 

```python
funds = pd.read_csv(data_file('weekly.csv'))
funds.head()
if funds.empty:
    raise ValueError('weekly.csv contains no observations. Use your original downloaded dataset.')
```

### Step 4.  What is the frequency of the dataset?

```python
dates = pd.DatetimeIndex(pd.to_datetime(funds['Date'])).sort_values()
pd.infer_freq(dates) if len(dates) >= 3 else 'Insufficient dates to infer frequency'
```

### Step 5. Set the column Date as the index.

```python
funds = funds.set_index('Date')
```

### Step 6. What is the type of the index?

```python
type(funds.index)
```

### Step 7. Set the index to a DatetimeIndex type

```python
funds.index = pd.to_datetime(funds.index)
funds = funds.sort_index()
```

### Step 8.  Change the frequency to monthly, sum the values and assign it to monthly.

```python
monthly = funds.resample('MS').sum(numeric_only=True, min_count=1)
monthly
```

### Step 9. You will notice that it filled the dataFrame with months that don't have any data with NaN. Let's drop these rows.

```python
monthly = monthly.dropna(how='all')
monthly
```

### Step 10. Good, now we have the monthly data. Now change the frequency to year.

```python
yearly = monthly.resample('YS').sum(min_count=1)
yearly
```
