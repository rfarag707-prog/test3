# Tips

### Step 1. Import the necessary libraries:

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv(data_file('tips.csv'))
tips.head()
```

### Step 4. Delete the Unnamed 0 column

```python
tips = tips.drop(columns=['Unnamed: 0'], errors='ignore')
```

### Step 5. Plot the total_bill column histogram

```python
sns.histplot(data=tips, x='total_bill', bins=20)
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.pairplot(tips[['total_bill', 'tip', 'size']])
plt.show()
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.stripplot(data=tips, x='day', y='total_bill', jitter=True)
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.stripplot(data=tips, x='tip', y='day', hue='sex', orient='h', jitter=True)
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(data=tips, x='day', y='total_bill', hue='time')
plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
sns.displot(data=tips, x='tip', col='time', col_order=['Dinner', 'Lunch'], bins=15)
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
sns.relplot(data=tips, x='total_bill', y='tip', hue='smoker', col='sex', col_order=['Male', 'Female'])
plt.show()
```

### BONUS: Create your own question and answer it using a graph.

```python
# How does the tip percentage vary by day?
tips.assign(tip_pct=tips['tip'].div(tips['total_bill'].replace(0, np.nan)).mul(100)).groupby('day')['tip_pct'].mean().plot.bar(ylabel='Mean tip percentage')
plt.tight_layout()
plt.show()
```
