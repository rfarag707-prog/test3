# Visualizing the Titanic Disaster

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Desaster/train.csv)

### Step 3. Assign it to a variable titanic 

```python
titanic = pd.read_csv(data_file('train.csv'))
titanic.head()
```

### Step 4. Set PassengerId as the index 

```python
titanic = titanic.set_index('PassengerId')
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
titanic['Sex'].value_counts().plot.pie(autopct='%1.1f%%', ylabel='', title='Passenger sex')
plt.show()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
sns.scatterplot(data=titanic, x='Fare', y='Age', hue='Sex')
plt.show()
```

### Step 7. How many people survived?

```python
titanic['Survived'].eq(1).sum()
```

### Step 8. Create a histogram with the Fare payed

```python
titanic['Fare'].plot.hist(bins=30, xlabel='Fare')
plt.show()
```

### BONUS: Create your own question and answer it.
