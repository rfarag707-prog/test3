### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Create a data dictionary that looks like the DataFrame below

```python
data = {'name': ['Bulbasaur', 'Charmander', 'Squirtle', 'Caterpie'], 'type': ['grass', 'fire', 'water', 'bug'], 'hp': [45, 39, 44, 45], 'evolution': ['Ivysaur', 'Charmeleon', 'Wartortle', 'Metapod'], 'pokedex': ['yes', 'no', 'yes', 'no']}
```

### Step 3. Assign it to a variable called pokemon
