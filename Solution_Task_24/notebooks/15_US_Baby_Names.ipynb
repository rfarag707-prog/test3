# US - Baby Names

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv). 

### Step 3. Assign it to a variable called baby_names.

```python
baby_names = pd.read_csv(data_file('US_Baby_Names_right.csv'))
baby_names.head()
```

### Step 4. See the first 10 entries

```python
baby_names.head(10)
```

### Step 5. Delete the column 'Unnamed: 0' and 'Id'

```python
baby_names = baby_names.drop(columns=['Unnamed: 0', 'Id'])
```

### Step 6. Is there more male or female names in the dataset?

```python
gender_summary = baby_names.groupby('Gender').agg(records=('Name', 'size'), distinct_names=('Name', 'nunique'), babies=('Count', 'sum'))
gender_summary
```

### Step 7. Group the dataset by name and assign to names

```python
names = baby_names.groupby('Name')[['Count']].sum()
names
```

### Step 8. How many different names exist in the dataset?

```python
len(names)
```

### Step 9. What is the name with most occurrences?

```python
names[names['Count'].eq(names['Count'].max())]
```

### Step 10. How many different names have the least occurrences?

```python
names['Count'].eq(names['Count'].min()).sum()
```

### Step 11. What is the median name occurrence?

```python
names['Count'].median()
```

### Step 12. What is the standard deviation of names?

```python
names['Count'].std()
```

### Step 13. Get a summary with the mean, min, max, std and quartiles.

```python
names.describe()
```
