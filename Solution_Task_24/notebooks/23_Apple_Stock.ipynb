# Apple Stock

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv)

```python
apple_path = data_file('appl_1980_2014.csv')
```

### Step 3. Assign it to a variable apple

```python
apple = pd.read_csv(apple_path)
apple.head()
```

### Step 4.  Check out the type of the columns

```python
apple.dtypes
```

### Step 5. Transform the Date column as a datetime type

```python
apple['Date'] = pd.to_datetime(apple['Date'])
```

### Step 6.  Set the date as the index

```python
apple = apple.set_index('Date')
```

### Step 7.  Is there any duplicate dates?

```python
apple.index.duplicated().any()
```

### Step 8.  Ops...it seems the index is from the most recent date. Make the first entry the oldest date.

```python
apple = apple.sort_index()
apple.head()
```

### Step 9. Get the last business day of each month

```python
# Keep the actual final observed trading date in each month.
monthly_last = apple.groupby(apple.index.to_period('M')).tail(1)
monthly_last
```

### Step 10.  What is the difference in days between the first day and the oldest

```python
(apple.index.max() - apple.index.min()).days
```

### Step 11.  How many months in the data we have?

```python
apple.index.to_period('M').nunique()
```

### Step 12. Plot the 'Adj Close' value. Set the size of the figure to 13.5 x 9 inches

```python
apple['Adj Close'].plot(figsize=(13.5, 9), ylabel='Adjusted close')
plt.show()
```

### BONUS: Create your own question and answer it.

```python
# What is the monthly percentage return?
monthly_last['Adj Close'].pct_change(fill_method=None).mul(100).plot(figsize=(12, 4), ylabel='Monthly return (%)')
plt.show()
```
