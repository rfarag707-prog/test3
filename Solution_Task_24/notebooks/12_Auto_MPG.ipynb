# MPG Cars

### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the first dataset [cars1](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv) and [cars2](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv).  

   ### Step 3. Assign each to a variable called cars1 and cars2

```python
cars1 = pd.read_csv(data_file('cars1.csv'))
cars2 = pd.read_csv(data_file('cars2.csv'))
```

### Step 4. Oops, it seems our first dataset has some unnamed blank columns, fix cars1

```python
cars1 = cars1.loc[:, ~cars1.columns.str.startswith('Unnamed')].copy()
cars1
```

### Step 5. What is the number of observations in each dataset?

```python
pd.Series({'cars1': len(cars1), 'cars2': len(cars2)})
```

### Step 6. Join cars1 and cars2 into a single DataFrame called cars

```python
cars = pd.concat([cars1, cars2], ignore_index=True)
cars
```

### Step 7. Oops, there is a column missing, called owners. Create a random number Series from 15,000 to 73,000.

```python
rng = np.random.default_rng(42)
owners = pd.Series(rng.integers(15000, 73001, size=len(cars)), name='owners')
owners
```

### Step 8. Add the column owners to cars

```python
cars['owners'] = owners
cars
```
