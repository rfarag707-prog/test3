### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv(data_file('chipotle.tsv'), sep='\t')
chipo.head()
```

### Step 4. See the first 10 entries

```python
chipo.head(10)
```

### Step 5. Create a histogram of the top 5 items bought

```python
top5 = chipo.groupby('item_name')['quantity'].sum().nlargest(5)
top5.plot.bar(xlabel='Item', ylabel='Quantity', title='Top 5 items bought')
plt.tight_layout()
plt.show()
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).str.strip().astype(float)
orders = chipo.groupby('order_id')[['quantity', 'item_price']].sum()
orders.plot.scatter(x='item_price', y='quantity', xlabel='Order price', ylabel='Items ordered')
plt.show()
```

### BONUS: Create a question and a graph to answer your own question.

```python
# Which items generate the most revenue?
chipo.groupby('item_name')['item_price'].sum().nlargest(5).plot.bar(ylabel='Revenue', title='Top 5 items by revenue')
plt.tight_layout()
plt.show()
```
