### Step 1. Import the necessary libraries

```python
from pathlib import Path
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_ROOT = Path(os.environ.get("PANDAS_DATA_ROOT", r"../pandas_exercises"))

def data_file(name):
    direct = DATA_ROOT / name
    if direct.is_file():
        return direct
    matches = list(DATA_ROOT.rglob(name)) if DATA_ROOT.exists() else []
    if len(matches) != 1:
        raise FileNotFoundError(f"Set DATA_ROOT to your extracted data folder. Expected one {name}; found {len(matches)}.")
    return matches[0]
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

### Step 9. Which was the most-ordered item? 

```python
item_totals = chipo.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
item_totals[item_totals.eq(item_totals.max())]
```

### Step 11. What was the most ordered item in the choice_description column?

```python
choice_totals = chipo.groupby('choice_description')['quantity'].sum()
choice_totals[choice_totals.eq(choice_totals.max())]
```

### Step 13. Turn the item price into a float

#### Step 13.a. Check the item price type

#### Step 13.b. Create a lambda function and change the type of item price

```python
to_float = lambda value: float(str(value).replace('$', '').strip())
chipo['item_price'] = chipo['item_price'].map(to_float)
```
