1. What is SQL and what is it used for?

```sql
-- SQL (Structured Query Language) is the language used to define, query, modify and control access to relational
-- data. It supports tasks such as finding records, creating tables and managing transactions.
```

2. What is the difference between SQL and MySQL?

```sql
-- SQL is a language; MySQL is a relational database management system that implements SQL. PostgreSQL and
-- SQL Server are other systems, with their own SQL extensions.
```

3. What are the different types of SQL commands?

```sql
-- Common teaching categories are DDL (CREATE, ALTER, DROP), DML (INSERT, UPDATE, DELETE), DQL
-- (SELECT), DCL (GRANT, REVOKE) and TCL (BEGIN, COMMIT, ROLLBACK). Some classifications include
-- SELECT within DML.
```

4. What is a database?

```sql
-- A database is an organized collection of data managed by a database system. A relational database stores
-- related data in tables and enforces rules through constraints.
```

5. What is a table in SQL?

```sql
-- A table is a named relation represented by columns and rows. Columns define attributes; each row represents
-- one record. Rows have no guaranteed order without ORDER BY.
```

6. What is a row in SQL?

```sql
-- A row is a single record containing values for the table's columns; for example, one employee with an ID, name
-- and salary.
```

7. What is a column in SQL?

```sql
-- A column is a named attribute with a data type, such as salary NUMERIC(12,2). It holds the same type of
-- information for all rows.
```

8. What is a primary key?

```sql
-- A primary key uniquely identifies each row and disallows NULL. A table has at most one primary key, which may
-- contain multiple columns.
```

9. What is a foreign key?

```sql
-- A foreign key constrains a column or column group to reference a qualifying unique or primary key in another
-- table, or the same table. Nullable foreign keys may represent no relationship.
```

10. What is a unique key?

```sql
-- A UNIQUE constraint prevents duplicate key values. A table may have several unique constraints. In
-- PostgreSQL, NULL values are distinct by default, so multiple NULLs are allowed unless NULLS NOT DISTINCT
-- is specified.
```

11. What is a composite key?

```sql
-- A composite key uses two or more columns together to identify a row, such as (employee_id, project_id) in an
-- assignment table.
```

12. What is a candidate key?

```sql
-- A candidate key is a minimal set of attributes that uniquely identifies a row: no subset remains unique. One
-- candidate is selected as the primary key; the others are alternate keys.
```

13. What is normalization?

```sql
-- Normalization organizes relations according to dependencies to reduce redundancy and avoid insertion, update
-- and deletion anomalies. It usually separates facts about different entities into related tables.
```

14. What are the different normal forms?

```sql
-- 1NF: attributes contain atomic values for the chosen model, with no repeating groups. 2NF: 1NF plus no non-key
-- attribute depending on only part of a candidate key. 3NF: for every nontrivial functional dependency X -> A, X is a
-- superkey or A is a prime attribute. BCNF: every nontrivial determinant is a superkey. 4NF removes independent
-- nontrivial multivalued dependencies unless the determinant is a superkey. 5NF removes nontrivial join
-- dependencies not implied by candidate keys.
```

15. What is denormalization?

```sql
-- Denormalization deliberately stores redundant or precomputed data to reduce expensive reads. Examples
-- include order totals or summary tables. It increases storage and requires a reliable method to keep copies
-- consistent.
```

16. What is the difference between DELETE, DROP, and TRUNCATE?

```sql
-- DELETE removes selected rows using WHERE (or all rows if omitted). TRUNCATE empties a whole table
-- without a row filter, keeping its definition. DROP removes the table itself. PostgreSQL allows all three to be rolled
-- back inside a transaction; TRUNCATE takes a strong lock and has different trigger behavior. Identity-reset and
-- rollback rules vary by database, so TRUNCATE is not universally non-transactional.
```

17. What is the difference between WHERE and HAVING?

```sql
-- WHERE filters input rows before grouping; HAVING filters groups after aggregation. Example: WHERE salary > 0
-- filters employees, while HAVING COUNT(*) > 5 filters department groups.
```

18. What is the difference between CHAR and VARCHAR?

```sql
-- CHAR(n) is fixed-length and blank-padded; VARCHAR(n) stores variable-length text up to n characters. Padding
-- and trailing-space comparison details depend on the database. CHAR is not automatically faster.
```

19. What is the difference between NULL and 0?

```sql
-- NULL means missing, unknown or not applicable. Zero is a known numeric value. Use IS NULL, not = NULL;
-- ordinary comparisons with NULL produce UNKNOWN.
```

20. What is the purpose of the SELECT statement?

```sql
-- SELECT retrieves or computes a result set from tables, views or expressions; it can filter, join, group and sort
-- data.
```

21. What is the purpose of the INSERT statement?

```sql
-- INSERT adds new rows to a table, either from explicit values or from a query result.
```

22. What is the purpose of the UPDATE statement?

```sql
-- UPDATE changes values in existing rows. A WHERE clause limits the affected rows; without it, all rows are
-- updated.
```

23. What is the purpose of the DELETE statement?

```sql
-- DELETE removes rows while preserving the table structure. WHERE selects which rows are removed.
```

24. What is the purpose of the CREATE statement?

```sql
-- CREATE defines a new database object, such as a table, view or index.
```

25. What is the purpose of the ALTER statement?

```sql
-- ALTER changes the definition of an existing object, for example by adding a column or a constraint.
```

26. What is the purpose of the DROP statement?

```sql
-- DROP removes a database object and its definition. Dependencies can prevent the operation unless explicitly
-- handled.
```

27. What is the purpose of the DISTINCT keyword?

```sql
-- DISTINCT removes duplicate combinations of the selected expressions from the result. It does not modify stored
-- rows.
```

28. What is the ORDER BY clause?

```sql
-- ORDER BY sorts the final result by one or more expressions. ASC is ascending and DESC is descending. Add a
-- unique tie-breaker for deterministic pagination.
```

29. What is the GROUP BY clause?

```sql
-- GROUP BY combines rows sharing grouping values into groups, usually for aggregate calculations such as SUM
-- or COUNT.
```

30. What is the LIMIT clause?

```sql
-- LIMIT restricts the number of returned rows in PostgreSQL and MySQL. Example: ORDER BY employee_id
-- LIMIT 10. SQL Server commonly uses TOP or OFFSET/FETCH.
```

31. What is the OFFSET clause?

```sql
-- OFFSET skips a specified number of result rows. Use it with a stable ORDER BY; large offsets can be inefficient
-- compared with keyset pagination.
```

32. What is the BETWEEN operator?

```sql
-- BETWEEN tests an inclusive range: x BETWEEN a AND b means x >= a AND x <= b. For timestamp periods, a
-- half-open range (>= start AND < next_start) often avoids end-of-day errors.
```

33. What is the IN operator?

```sql
-- IN tests membership in a value list or single-column subquery. Example: department_id IN (10, 20). A NULL
-- inside NOT IN can cause UNKNOWN rather than TRUE.
```

34. What is the LIKE operator?

```sql
-- LIKE matches text patterns: % means any sequence of characters and _ means one character. PostgreSQL
-- LIKE is case-sensitive; ILIKE is a PostgreSQL case-insensitive extension.
```

35. What is the IS NULL operator?

```sql
-- IS NULL tests whether a value is NULL. IS NOT NULL checks that a value is present.
```

36. What is the EXISTS operator?

```sql
-- EXISTS returns TRUE if its subquery returns at least one row, regardless of the values selected; otherwise it
-- returns FALSE.
```

37. What is the ANY operator?

```sql
-- A comparison with ANY(subquery) is TRUE if at least one comparison is TRUE. Example: salary > ANY(...)
-- means higher than at least one returned salary. An empty set gives FALSE; NULLs can produce UNKNOWN.
-- 
```

38. What is the ALL operator?

```sql
-- A comparison with ALL(subquery) is TRUE if all comparisons are TRUE. An empty set gives TRUE; NULLs can
-- produce UNKNOWN. With a nonempty, non-NULL set, salary > ALL(...) means higher than its maximum.
```

39. What is the UNION operator?

```sql
-- UNION combines compatible query results into one set and removes duplicate rows. Inputs must return the same
-- number of columns with compatible types.
```

40. What is the difference between UNION and UNION ALL?

```sql
-- UNION removes duplicate output rows. UNION ALL preserves every row, including duplicates, and avoids
-- duplicate-elimination work. Neither guarantees output order without a final ORDER BY.
```

71. What is a JOIN in SQL?

```sql
-- A JOIN combines rows from two table expressions based on a relationship or condition. A match may produce
-- multiple output rows when the relationship is one-to-many.
```

72. What is an INNER JOIN?

```sql
-- INNER JOIN returns only pairs of rows that satisfy the join condition.
```

73. What is a LEFT JOIN?

```sql
-- LEFT JOIN returns matching pairs plus every unmatched left-side row, with NULL values for the right-side
-- columns.
```

74. What is a RIGHT JOIN?

```sql
-- RIGHT JOIN returns matching pairs plus every unmatched right-side row, with NULL values for the left-side
-- columns.
```

75. What is a FULL OUTER JOIN?

```sql
-- FULL OUTER JOIN preserves matches and unmatched rows from both sides. Missing-side columns are NULL.
-- PostgreSQL supports it directly; MySQL does not have a native FULL OUTER JOIN.
```

76. What is a CROSS JOIN?

```sql
-- CROSS JOIN produces the Cartesian product: each row of one input is combined with every row of the other. For
-- m and n rows, the output has m * n rows.
```

77. What is a SELF JOIN?

```sql
-- A self join joins a table to itself using distinct aliases. It is useful for employee-manager relationships or
-- comparing records within a table.
```

78. What is the difference between INNER JOIN and LEFT JOIN?

```sql
-- INNER JOIN discards unmatched rows. LEFT JOIN preserves unmatched rows from the left input. Filtering
-- right-side columns in WHERE can accidentally remove the preserved rows.
```

79. What is the difference between LEFT JOIN and RIGHT JOIN?

```sql
-- LEFT JOIN preserves the left input; RIGHT JOIN preserves the right input. Swapping the table positions can
-- express the same relationship using the other join type.
```

91. What are aggregate functions in SQL?

```sql
-- Aggregate functions summarize a set of rows into a value, such as a count, sum or average. With GROUP BY
-- they return one result per group. Most ignore NULL inputs; COUNT(*) counts all rows.
```

92. What is the COUNT() function?

```sql
-- COUNT(*) counts rows. COUNT(column) counts non-NULL values. COUNT(DISTINCT column) counts distinct
-- non-NULL values. COUNT returns zero for an empty input.
```

93. What is the SUM() function?

```sql
-- SUM(expression) adds non-NULL numeric inputs. With no non-NULL inputs, it returns NULL; use
-- COALESCE(SUM(...), 0) only if zero is the intended business meaning.
```

94. What is the AVG() function?

```sql
-- AVG(expression) is the arithmetic mean of non-NULL inputs: their sum divided by their count. NULL values are
-- excluded, not treated as zero.
```

95. What is the MAX() function?

```sql
-- MAX(expression) returns the greatest non-NULL input, or NULL when there is none. It does not automatically
-- return the rest of the row containing that value.
```

96. What is the MIN() function?

```sql
-- MIN(expression) returns the smallest non-NULL input, or NULL when there is none.
```

112. What is a correlated subquery?

113. What is the difference between correlated and non-correlated subqueries?

131. What is a view in SQL?

```sql
-- A regular view is a stored query exposed like a table. It normally stores the query definition, not its result. It can
-- simplify reusable queries and provide controlled access to selected data.
```

132. What is an index in SQL?

```sql
-- An index is an auxiliary structure that helps locate rows or satisfy ordering without scanning all data. It consumes
-- storage and adds maintenance work to writes; an index does not guarantee a query will use it.
```

133. What are clustered and non-clustered indexes?

```sql
-- In engines such as SQL Server, a clustered index organizes table data by its key; a non-clustered index is a
-- separate structure containing keys and row locators. PostgreSQL uses heap tables with separate indexes. Its
-- CLUSTER command physically reorders a table once, but later writes do not maintain that order. Never assume
-- result order without ORDER BY.
```

134. What is a stored procedure?

```sql
-- A stored procedure is a named routine stored in the database and invoked with CALL in PostgreSQL. It
-- encapsulates server-side work and may accept parameters. Transaction-control rules depend on the database
-- and calling context.
```

135. What is a trigger?

```sql
-- A trigger automatically invokes database logic when an event occurs, such as INSERT, UPDATE or DELETE. It
-- can audit changes or enforce rules that need logic beyond ordinary constraints.
```

136. What is a cursor?

```sql
-- A cursor provides controlled access to a query result, often fetching rows incrementally. It is useful for streaming
-- or procedural processing; set-based SQL is usually preferable for bulk transformations.
```

137. What is a transaction?

```sql
-- A transaction is a logical unit of database work that is committed as a whole or rolled back. Example: subtracting
-- money from one account and adding it to another in one transaction.
```

138. What are ACID properties?

```sql
-- Atomicity: all-or-nothing changes. Consistency: committed transactions preserve defined integrity rules when
-- application logic is correct. Isolation: concurrent work observes the guarantees of the chosen isolation level.
-- Durability: committed changes survive failures according to the configured durability guarantees.
```

139. What is a savepoint?

```sql
-- A savepoint marks a position inside a transaction. ROLLBACK TO SAVEPOINT can undo work after that point
-- without discarding all earlier work in the transaction.
```

140. What is a rollback?

```sql
-- ROLLBACK abandons uncommitted changes in a transaction. ROLLBACK TO SAVEPOINT undoes only work
-- since the named savepoint.
```

141. What is a commit?

```sql
-- COMMIT successfully ends a transaction and makes its changes durable and visible according to isolation and
-- durability settings. Deferred constraint checks may cause a commit attempt to fail.
```

142. What is deadlock?

```sql
-- A deadlock occurs when transactions wait on resources held by each other in a cycle. The database typically
-- aborts one transaction. Use consistent lock ordering, short transactions and bounded retry logic.
```

143. What is database locking?

```sql
-- Locking coordinates concurrent access to resources such as rows and tables. Different lock modes protect reads
-- or modifications. MVCC allows many reads without blocking writes, but does not eliminate all locks.
```

144. What is isolation level?

```sql
-- An isolation level specifies which effects of concurrent transactions can be observed. Standard levels are Read
-- Uncommitted, Read Committed, Repeatable Read and Serializable. In PostgreSQL, Read Uncommitted behaves
-- like Read Committed; Serializable transactions may need retries.
```

145. What is a sequence in SQL?

160. Write a query using window functions.

```sql
SELECT employee_id, department_id, salary,
AVG(salary) OVER (
PARTITION BY department_id
) AS department_average,
SUM(salary) OVER (
PARTITION BY department_id ORDER BY employee_id
ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
) AS running_salary
FROM employees
ORDER BY department_id NULLS LAST, employee_id;
```

162. What is an ER diagram?

163. What is cardinality?

164. What is a one-to-one relationship?

```sql
-- A one-to-one relationship allows at most one related row on each side, such as Person and PassportRecord in a
-- simplified model. A UNIQUE foreign key can enforce the at-most-one rule; NOT NULL expresses required
-- participation on the referencing side.
```

165. What is a one-to-many relationship?

166. What is a many-to-many relationship?

167. What is referential integrity?

```sql
-- Referential integrity ensures that foreign-key references identify existing referenced keys, unless a permitted
-- NULL means no reference. Deletion/update actions may restrict changes or propagate them through CASCADE,
-- SET NULL or other supported rules.
```

168. What is a surrogate key?

```sql
-- A surrogate key is an artificial identifier without business meaning, such as an identity integer or UUID. It offers a
-- stable reference, while business uniqueness may still need a separate constraint.
```

169. What is a natural key?

```sql
-- A natural key comes from business data, such as a standardized country code. It must be unique and stable for
-- the modeled entity; names and email addresses should not be assumed immutable.
```

170. What are database constraints?

```sql
-- Constraints are database-enforced rules such as PRIMARY KEY, FOREIGN KEY, UNIQUE, NOT NULL and
-- CHECK. They protect integrity even when multiple applications write to the same tables.
```

171. What is the NOT NULL constraint?

```sql
-- NOT NULL prevents a column from storing NULL. It does not prevent an empty string or zero; add other rules if
-- those are invalid.
```

172. What is the CHECK constraint?

```sql
-- CHECK requires a condition not to be FALSE. In PostgreSQL a NULL/UNKNOWN check result passes, so
-- CHECK(salary >= 0) still allows NULL unless NOT NULL is also specified.
```

173. What is the DEFAULT constraint?

```sql
-- DEFAULT supplies a value when an INSERT omits a column or uses DEFAULT. It does not replace an explicitly
-- supplied NULL. PostgreSQL treats this as a default expression rather than an integrity constraint in the same
-- sense as CHECK.
```

174. What is the purpose of indexing?

```sql
-- Indexing reduces work for suitable filters, joins and ordering; unique indexes can also support uniqueness
-- enforcement. The trade-offs are disk space, write overhead and maintenance.
```

175. How does normalization improve database design?

```sql
-- Normalization stores each fact in an appropriate place, reduces inconsistent copies and prevents common
-- insertion, update and deletion anomalies. Keys and constraints connect the resulting tables.
```

176. What are the disadvantages of normalization?

```sql
-- Highly normalized schemas can require more joins and more complex reporting queries. Some read-heavy
-- workloads benefit from carefully maintained summaries or denormalized models after measuring the bottleneck.
```

177. What is schema design?

```sql
-- Schema design defines tables, columns, types, keys, constraints and relationships to support business rules and
-- workloads. It also considers indexes, lifecycle, security and migration strategy.
```

178. What is star schema?

```sql
-- A star schema has a central fact table at a clearly defined grain, linked to relatively denormalized dimensions
-- such as date, product and customer. It is common in analytical reporting.
```

179. What is snowflake schema?

```sql
-- A snowflake schema normalizes dimensions into related sub-tables, such as Product -> Category. This reduces
-- repeated dimension data but can require more joins than a star schema.
```

180. What is data warehousing?

```sql
-- Data warehousing consolidates historical data from operational sources for analysis. ETL/ELT pipelines clean
-- and integrate data into analytical models, typically optimized for large reads and consistent reporting.
```

181. How would you optimize a slow SQL query?

```sql
-- First reproduce the slow query with representative parameters and data. Inspect EXPLAIN and, where safe,
-- EXPLAIN (ANALYZE, BUFFERS). Compare estimates with actual rows, look for large scans, bad joins, sorts,
-- spills or lock waits, then address the measured cause. Try suitable indexes, selective predicates, corrected joins
-- and updated statistics; remeasure the result. EXPLAIN ANALYZE executes the query.
```

182. How would you identify duplicate records?

```sql
-- Define which business columns make records duplicates, then group by those columns and use HAVING
-- COUNT(*) > 1. Use ROW_NUMBER to identify excess copies. Q106 checks email duplicates; Q62 checks full
-- staging-row duplicates. Never assume identical names identify one person.
```

183. How would you improve database performance?

```sql
-- Measure query latency, throughput, waits and resource saturation. Improve high-impact queries and indexes,
-- keep planner statistics and PostgreSQL vacuum maintenance healthy, use connection pooling, and reduce long
-- transactions. Add caching, replicas, partitioning or resources only when workload evidence supports them.
```

184. How would you handle large datasets in SQL?

```sql
-- Filter early where logically valid, select needed columns, use suitable indexes and process writes in bounded
-- batches. Use bulk loading for large imports and keyset pagination for long result lists. Partition when pruning or
-- retention benefits justify it; avoid loading the entire result into application memory.
```

185. What are the best practices for writing SQL queries?

```sql
-- Use explicit columns, clear aliases, readable formatting and correct keys in joins. Bind parameters, handle NULL
-- explicitly, use deterministic ORDER BY for pagination, and define transaction boundaries. Check affected rows
-- for updates/deletes and validate performance on representative data.
```

186. What is query optimization?

```sql
-- Query optimization chooses or improves an execution strategy that produces the same correct result with less
-- work. It includes join ordering, access paths, predicate handling and cost estimation.
```

187. What is execution plan?

```sql
-- An execution plan describes operators used to run a query, such as sequential scans, index scans, joins, sorts
-- and aggregates. Estimated plans predict rows and costs; actual execution statistics show observed work and
-- timing. Cost units are not milliseconds.
```

188. What is indexing strategy?

```sql
-- Choose indexes from real filter, join and sort patterns. For common B-tree cases, equality columns followed by
-- useful range/order columns are a starting point, not a universal rule. Consider selectivity, composite order,
-- partial/covering indexes and write costs. Measure usage and remove redundant indexes carefully.
```

189. How do you secure a database?

```sql
-- Use least-privilege roles, strong authentication, restricted network access and TLS. Patch the database, protect
-- credentials, encrypt backups, audit sensitive operations and separate application access from administration.
-- Use parameter binding and, where appropriate, row-level access controls.
```

```sql
-- Use parameterized queries so values are not concatenated into SQL text. Allow-list dynamic identifiers or sort
-- directions, which value parameters cannot represent. Apply least privilege; escaping input alone is not the
-- preferred defense.
```

193. What is database partitioning?

```sql
-- Partitioning splits one logical table into smaller physical partitions, for example by date range or hash. It can
-- improve pruning and data-retention operations, but does not automatically accelerate queries that need all
-- partitions.
```

194. What is sharding?

```sql
-- Sharding distributes different subsets of data across separate database nodes. It can scale storage and writes,
-- but complicates cross-shard joins, transactions, rebalancing and globally unique identifiers.
```

195. What is replication?

```sql
-- Replication copies database changes to other instances for availability or read scaling. Synchronous and
-- asynchronous modes have different latency and durability trade-offs. Replicas can lag, and copied deletions
-- make replication insufficient as a backup.
```

196. What is database migration?

```sql
-- A database migration is a controlled, versioned change to schema or data. Test it, estimate locks and runtime,
-- plan rollback or forward repair, and use backward-compatible expand/migrate/contract steps when applications
-- must remain available.
```

197. How do you monitor SQL performance?

```sql
-- Track query latency percentiles, throughput, errors, locks, connections, CPU, I/O, storage and replication lag. In
-- PostgreSQL, use logs, pg_stat_activity, pg_stat_statements when enabled, and execution plans. Alert against
-- baselines and service objectives, then investigate specific regressions.
```
