## Part 1: Research

### 1. What is Pandas, and why is it used in data analysis?
Pandas is a Python library used to organize, clean, explore, and analyze structured data. It provides convenient tools for working with tables, missing values, filtering, grouping, and statistical summaries.

### 2. What is a Pandas Series?
A Series is a one-dimensional labeled array. It stores values and an index that identifies each value.

### 3. What is a Pandas DataFrame?
A DataFrame is a two-dimensional table made of rows and columns. Each column can have its own data type, and both rows and columns have labels.

### 4. What is the difference between `loc[]` and `iloc[]`?
`loc[]` selects data by row and column labels. `iloc[]` selects data by integer positions, starting at zero.

### 5. Explain the purpose of `head()`, `tail()`, `info()`, and `describe()`.
- `head()` returns the first rows (five by default).
- `tail()` returns the last rows (five by default).
- `info()` displays column names, non-null counts, data types, and memory information.
- `describe()` provides summary statistics for numeric columns, such as count, mean, standard deviation, minimum, quartiles, and maximum.


## Part 2: Practical Application

Create a DataFrame with 10 records, inspect it, select rows and columns, and summarize its numeric data.

```python
import pandas as pd

data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Ali', 'Mona', 'Omar', 'Sara', 'Youssef', 'Nour', 'Hana', 'Adam', 'Laila', 'Ziad'],
    'Age': [20, 21, 19, 22, 20, 23, 21, 19, 22, 20],
    'Department': ['AI', 'Data Science', 'AI', 'Networks', 'Data Science', 'AI', 'Networks', 'AI', 'Data Science', 'Networks'],
    'Grade': [88, 92, 79, 95, 84, 90, 86, 81, 94, 87]
}

df = pd.DataFrame(data)
df
```

### 2. Display the first five and last five records

```python
df.head()
```

```python
df.tail()
```

### 3. Display the DataFrame's shape, column names, and data types

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
print('Data types:\n', df.dtypes)
```

### 4. Select specific rows and columns using `loc[]` and `iloc[]`

```python
# loc[] uses labels
df.loc[0:3, ['Name', 'Department', 'Grade']]
```

```python
# iloc[] uses integer positions
df.iloc[0:4, [1, 3, 4]]
```

### 5. Generate and interpret a statistical summary using `describe()`

```python
df.describe()
```

### Interpretation
- `count` shows how many non-missing numeric values are present in each column.
- `mean` is the average value.
- `std` measures how spread out the values are around the mean.
- `min` and `max` are the smallest and largest values.
- `25%`, `50%`, and `75%` are the first quartile, median, and third quartile.
